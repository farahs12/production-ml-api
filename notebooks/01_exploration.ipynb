```python
df =df.rename(columns={"default.payment.next.month": "default", "PAY_0":"PAY_1"})
df = df.drop(columns=["ID"])
df["EDUCATION"]=df["EDUCATION"].replace({0:4, 5:4, 6:4})
df["MARRIAGE"]=df["MARRIAGE"].replace({0 : 3})
```

```python
print(df.shape)
print(df.columns.tolist())
print(df["EDUCATION"].value_counts())
print(df["MARRIAGE"].value_counts())
```

```python
from sklearn.model_selection import train_test_split
train_df, temp_df = train_test_split(df, test_size=0.3 , stratify=df["default"], random_state=42)
val_df, test_df = train_test_split(temp_df, test_size=0.5, stratify= temp_df["default"], random_state= 42)
```

```python
for name, part in [("train", train_df), ("val", val_df), ("test", test_df)]:
    print(name, part.shape, round(part["default"].mean(), 3))
```

```python
X_train = train_df.drop(columns=["default"])
y_train = train_df["default"]
X_val = val_df.drop(columns=["default"])
y_val = val_df["default"]
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
train_df.info()
```

```python
categorical_cols = ["SEX","EDUCATION","MARRIAGE"]
numeric_cols = X_train.columns.drop(categorical_cols).tolist()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder

scaler = StandardScaler()
encoder = OneHotEncoder(handle_unknown = "ignore")
```

```python
from sklearn.compose import ColumnTransformer

numeric_step = ("num" , scaler , numeric_cols)
categorical_step = ("cat" , encoder, categorical_cols)

preprocessor = ColumnTransformer([numeric_step, categorical_step])
```

```python
preprocessor.fit(X_train)

X_train_ready = preprocessor.transform(X_train)
print(X_train_ready.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
log_reg = LogisticRegression(max_iter=1000)
```

```python
model= Pipeline([("preprocessor", preprocessor),("classifier", log_reg)])
```

```python
model
```

```python
model.fit(X_train, y_train)
```

```python
y_val_pred = model.predict(X_val)
```
